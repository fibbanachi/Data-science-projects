```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    f1_score,
    classification_report,
    confusion_matrix,
    roc_auc_score
)
from catboost import CatBoostClassifier
from tqdm import tqdm
```

```python
train, test = train_test_split(df, train_size = 0.6, random_state = 42)
```

```python
train, val = train_test_split(train, train_size = 0.8, random_state = 42)
```

```python
df.info()
```

```python
df['Embarked']
```

```python
X = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch','Fare','Embarked']
cat_features = ['Sex', 'Pclass', 'Embarked', 'SibSp', 'Parch']
Y = ['Survived']
```

```python
model = CatBoostClassifier(cat_features=cat_features,
                          eval_metric='AUC',
                          random_seed=42,
                          verbose=100)
```

```python
model.fit(train[X],train[Y], eval_set=(val[X],val[Y]))
```

```python
y_pred_val = model.predict(val[X])
```

```python
y_pred_proba = model.predict_proba(val[X])[:, 1]
```

```python
y_valid = val[Y]
```

```python
print("F1-score:", f1_score(y_valid, y_pred_val))
print("ROC-AUC:", roc_auc_score(y_valid, y_pred_proba))
print("\nПолный отчёт:")
print(classification_report(y_valid, y_pred_val))
print("Матрица ошибок:")
print(confusion_matrix(y_valid, y_pred_val))
```

```python
y_test = test[Y]
```

```python
y_pred_test = model.predict(test[X])
```

```python
y_pred_proba = model.predict_proba(test[X])[:, 1]
```

```python
print("F1-score:", f1_score(y_test, y_pred_test))
print("ROC-AUC:", roc_auc_score(y_test, y_pred_proba))
print("\nПолный отчёт:")
print(classification_report(y_test, y_pred_test))
print("Матрица ошибок:")
print(confusion_matrix(y_test, y_pred_test))
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import LabelEncoder
```

```python
features = ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch','Fare','Embarked']
cat_features = ['Sex', 'Embarked']
```

```python
Y_train = train['Survived']
X_train = train[features].copy()
```

```python
X_train_split, X_val, y_train_split, y_val = train_test_split(
    X_train, y_train,
    train_size=0.8,
    random_state=42,
    stratify=y_train  # Важно для Titanic!
)
```

```python
le_sex = LabelEncoder()
le_embarked = LabelEncoder()

X_train_split['Sex'] = le_sex.fit_transform(X_train_split['Sex'])
X_val['Sex'] = le_sex.transform(X_val['Sex'])

X_train_split['Embarked'] = le_embarked.fit_transform(X_train_split['Embarked'])
X_val['Embarked'] = le_embarked.transform(X_val['Embarked'])
```

```python
model = RandomForestClassifier(n_estimators=200, max_depth=7, random_state=42)
model.fit(X_train_split, y_train_split)
```

```python
y_pred_train = model.predict(X_train_split)
y_pred_val = model.predict(X_val)
```

```python
print("\n=== CLASSIFICATION REPORT (Val) ===")
print(classification_report(y_val, y_pred_val))
```

```python
y_prob_train = model.predict_proba(X_train_split)[:, 1]
y_prob_val = model.predict_proba(X_val)[:, 1]
```

```python
print("=== ПОЛНЫЙ ОТЧЕТ (Validation) ===")
print(f"F1-score:  {f1_score(y_val, y_pred_val):.3f}")
print(f"ROC-AUC:   {roc_auc_score(y_val, y_prob_val):.3f}")
print()
```
